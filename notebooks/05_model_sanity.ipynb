# 05 - War Room model sanity
Runs the engine and optimizer on real congressional districts and checks speed and diminishing returns.


```python
import time
from campaign_intel.data.snapshot import read_table
from campaign_intel.model.assumptions import load_defaults
from campaign_intel.model.engine import RaceContext, response_curve, simulate
from campaign_intel.model.optimizer import optimize

units = read_table("units_cd").assign(opp_share=0.4)
a = load_defaults()
ctx = RaceContext("statewide", "cd", None, 8, 2_000_000, units)
t = time.time(); plan, diff = optimize(None, ctx, a); print(f"optimizer {time.time() - t:.2f}s")
r = simulate(plan, ctx, a)
print(f"win {r.baseline_win_prob:.1%} -> {r.win_prob:.1%}; net votes {r.total_delta_votes:,.0f}; $/vote {r.cost_per_net_vote:,.0f}")
```

```python
# Diminishing returns in one district
response_curve(None, ctx, a, "CD1", "persuasion", 8, 300_000, 7).pivot(index="extra_dollars", columns="channel", values="delta_votes").round(0)
```
