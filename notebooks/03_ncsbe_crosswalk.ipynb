# 03 - NCSBE precinct crosswalk checks
Verifies that districts and counties built from NCSBE precinct files reconcile to statewide totals.


```python
from campaign_intel.data.snapshot import read_table

sw = read_table("statewide").iloc[0]
for geo in ["cd", "sd", "hd", "county"]:
    u = read_table(f"units_{geo}")
    print(geo, len(u), u["total_votes_2024"].sum() - sw["pres24_total"], u["registered_voters"].sum() - sw["registered_2024"])
```

```python
# County <-> district overlap (share of each county's registered voters in each district)
xw = read_table("county_district_crosswalk")
xw[(xw.unit_type == "cd") & (xw.share_of_county < 0.99)].sort_values("county_fips").head(20)
```

```python
# Baseline vs 2026 actual by congressional district
u = read_table("units_cd")
u.assign(swing=u.actual26_margin_pct - u.dem_margin_pct)[["unit_id", "dem_margin_pct", "actual26_margin_pct", "swing"]].round(1)
```
