# 01 - Google Political Ads: NC snapshot EDA
Explores the Google tables pulled by `campaign_intel.ingest.google_ads` (run the snapshot builder first).


```python
import pandas as pd
from campaign_intel.data.snapshot import read_table

cr = read_table("google_creatives")
print(len(cr), "NC-targeted creatives;", f"{cr['nc_only'].mean():.0%} target only NC")
cr.groupby("ad_type")[["nc_spend_low", "nc_spend_mid", "nc_spend_high"]].sum().round(-3)
```

```python
# Weekly spend of NC-focused advertisers (>= 90% of lifetime US spend in NC) - the clean NC time series
wk = read_table("google_weekly_nc_focused")
wk.groupby("week_start_date")["spend_usd"].sum().tail(12)
```

```python
# Google's own congressional-district spend; labels carry the map vintage
read_table("google_cd_spend_units").pivot_table(index="unit_id", columns="map_vintage", values="spend_usd", aggfunc="sum")
```
